```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth
from itertools import combinations

df = pd.read_csv('vitrina_clients_clusters_all_new (3).csv')
df.set_index('Телефон_new', inplace=True)
```

```python
df['K9_Кластер'].value_counts()
```

```python
cat_cols = [c for c in df.columns if c.startswith('G2_')]
df_cat = df[cat_cols + ['K9_Кластер']].copy()
```

```python
def analyze_cluster(df_sub, cat_cols,
                    min_support=0.03,
                    lift_threshold=0.8):
    basket = df_sub[cat_cols].gt(0).astype(int)
    mandatory_cols = basket.columns[basket.sum(axis=0) == len(basket)]
    if len(mandatory_cols):
        print("Dropped mandatory category(ies):", ", ".join(mandatory_cols))
        basket = basket.drop(columns=mandatory_cols)
    else:
        print("No mandatory category found.")
    basket = basket[basket.sum(axis=1) >= 2]

    freq_itemsets = apriori(
        basket,
        min_support=min_support,
        use_colnames=True
    )
    rules = association_rules(
        freq_itemsets,
        metric="lift",
        min_threshold=lift_threshold
    )

    rules = rules.sort_values('lift', ascending=False).reset_index(drop=True)
    return freq_itemsets, rules
```

```python
def analyze_cluster_fpgrowth(df_sub, cat_cols,
                             min_support=0.03,
                             lift_threshold=0.8):
    basket = df_sub[cat_cols].gt(0).astype(int)

    mandatory_cols = basket.columns[basket.sum(axis=0) == len(basket)]
    basket = basket.drop(columns=mandatory_cols, errors='ignore')

    basket = basket[basket.sum(axis=1) >= 2]

    freq_itemsets = fpgrowth(
        basket,
        min_support=min_support,
        use_colnames=True
    )

    rules = association_rules(
        freq_itemsets,
        metric="lift",
        min_threshold=lift_threshold
    ).sort_values('lift', ascending=False).reset_index(drop=True)

    return freq_itemsets, rules
```

```python
from itertools import combinations
import pandas as pd

def eclat_pairs(df_sub, cat_cols, min_support=0.03):
    basket = df_sub[cat_cols].gt(0)
    n = len(basket)

    # защита от пустого кластера
    if n == 0:
        return pd.DataFrame(columns=['itemset', 'support'])

    # vertical (TID) lists
    tid = {
        col: set(basket.index[basket[col]])
        for col in basket.columns
        if basket[col].sum() / n >= min_support
    }

    itemsets = []
    for (i1, t1), (i2, t2) in combinations(tid.items(), 2):
        inter = t1 & t2
        sup = len(inter) / n
        if sup >= min_support:
            itemsets.append({'itemset': frozenset([i1, i2]), 'support': sup})

    if not itemsets:
        return pd.DataFrame(columns=['itemset', 'support'])

    return (pd.DataFrame(itemsets)
              .sort_values('support', ascending=False)
              .reset_index(drop=True))
```

```python
all_results = {}
for cluster_val in sorted(df_cat['K9_Кластер'].unique()):
    sub = df_cat[df_cat['K9_Кластер'] == cluster_val]
    freq, rules = analyze_cluster(sub, cat_cols)
    all_results[cluster_val] = {
        'freq_itemsets': freq,
        'rules': rules
    }
```

```python
fp_results = {}
for cluster_val in sorted(df_cat['K9_Кластер'].unique()):
    sub = df_cat[df_cat['K9_Кластер'] == cluster_val]
    freq, rules = analyze_cluster_fpgrowth(sub, cat_cols)
    fp_results[cluster_val] = {'freq_itemsets': freq, 'rules': rules}
```

```python
eclat_results = {}
for cluster_val in sorted(df_cat['K9_Кластер'].unique()):
    sub = df_cat[df_cat['K9_Кластер'] == cluster_val]
    pairs = eclat_pairs(sub, cat_cols, min_support=0.03)
    eclat_results[cluster_val] = pairs
```
